## 实验 1：中文分词（jieba）
目标：把一句中文切成词语列表，观察不同分词模式的差异。


```python
# 第 1 行：导入中文分词库 jieba
import jieba

# 第 2 行：准备一段待分词的中文文本
text = "自然语言处理是人工智能的重要领域"

# 第 3 行：用 jieba.cut 进行“精确模式”分词（默认）。返回一个可迭代的生成器对象
# 第 4 行：用 list(...) 把生成器转成列表，方便直接查看
# 第 5 行：print 打印结果，前缀"分词结果："便于阅读
print("分词结果：", list(jieba.cut(text)))
```

> 说明：`jieba.cut` 默认是“精确模式”，尽量把句子切得最合理，适合文本分析。
> 如果想用于搜索引擎，可以用“搜索引擎模式”，会切出更多重叠词。


```python
# 第 1 行：再准备一句话做模式对比
sentence = "苏州城市学院的自然语言处理课程很有趣"

# 第 2 行：精确模式（默认）
precise = list(jieba.cut(sentence))

# 第 3 行：搜索引擎模式（切出更多重叠词，便于索引/搜索）
search_mode = jieba.lcut_for_search(sentence)

# 第 4-5 行：打印两种模式的分词结果做对比
print("精确模式：", precise)
print("搜索模式：", search_mode)

# 同学们可以试试分词“南京市长江大桥”
# 在“精确模式”基础上，搜索引擎模式再额外切出一些更细的词，产生“重叠分词”，方便搜索引擎建立索引。
```

## 实验 2：情感分析
目标：判断一句话的情绪倾向（正向/负向）。

为了节约时间，在运行之前，需要先在Hugging Face里面手动下载模型！

网址为：https://huggingface.co/tabularisai/multilingual-sentiment-analysis/tree/main

下载后，放置在一个名为multilingual-sentiment-analysis的文件夹中。ps.该模型最好放置在一个全英文路径的位置，防止后续运行报错！


```python
# 第 1 行：从 transformers 中导入 pipeline 构造器
from transformers import pipeline


# 第 2-6 行：创建“情感分析”流水线（pipeline）
#            指定中文二分类情感模型：uer/roberta-base-finetuned-jd-binary-chinese
#            第一次运行会联网下载模型；之后会使用本地缓存，加速很多
classifier = pipeline("text-classification", model=r"C:\Users\18378\Desktop\multilingual-sentiment-analysis")  #这里是我下载的模型所放置位置的路径，请自行修改！！！

# 第 7 行：准备想要分析的中文句子
text = "这家餐厅的味道很好，但是服务态度太差了。"

# 第 8-9 行：把文本输入 pipeline，得到列表；取第 0 个字典元素作为结果
result = classifier(text)[0]

# 第 10-12 行：打印原句、预测标签（label）和置信度（score）
print("原句：", text)
print("预测标签：", result["label"])          # 可能是 "positive" 或 "negative"
print("置信度：", round(float(result["score"]), 4))
```
